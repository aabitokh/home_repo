[ 7 kyu Disemvowel Trolls](
https://www.codewars.com/kata/52fba66badcd10859f00097e/train/python)

```python
def disemvowel(string_):
  vowels = ['a', 'o', 'e', 'i', 'u']
  return ''.join([word for word in list(string_.lower()) if word not in vowels])
```

```python
string = 'LOL suck'
```

```python
disemvowel(string)
```

[7 kyu Growth of a Population](https://www.codewars.com/kata/563b662a59afc2b5120000c6/train/python)

```python
def nb_year(p0, percent, aug, p):

  counter = 0
  print(p0, p )
  while p0 < p:
    p0 = int(p0 + p0 * percent/100 + aug)
    print('new p0', p0)
    counter += 1
    print(counter, 'years')

  return counter
```

```python
nb_year(1000, 2, 50, 1200)
```

[7 kyu Get the Middle Character](https://www.codewars.com/kata/56747fd5cb988479af000028/train/python)

```python
def get_middle(s):
    lenght_w = len(s)
    if lenght_w%2 == 0:
        return s[int(lenght_w/2-1)] + s[int(lenght_w/2)]
    else:
        return s[int(lenght_w/2)]
```
